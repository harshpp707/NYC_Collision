```python
import pandas as pd

```

```python
df = pd.read_csv("data/nyc_collision.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df['crash_date'] = pd.to_datetime(df['crash_date'], errors='coerce')
df['crash_time'] = pd.to_datetime(df['crash_time'], format='%H:%M', errors='coerce')
```

```python
distinct_boroughs = df['borough'].unique()
print(distinct_boroughs)
```

```python
df = df.dropna(subset=['borough', 'latitude', 'longitude'])
```

```python
df.head()
```

```python
distinct_boroughs = df['borough'].unique()
print(distinct_boroughs)
```

```python
df['hour'] = df['crash_time'].dt.hour
```

```python
def time_of_day(h):
    if 5 <= h < 12:
        return 'Morning'
    elif 12 <= h < 17:
        return 'Afternoon'
    elif 17 <= h < 21:
        return 'Evening'
    else:
        return 'Night'
```

```python
df['time_of_day'] = df['hour'].apply(time_of_day)
```

```python
ped = df[df['number_of_pedestrians_injured'] + df['number_of_pedestrians_killed'] > 0]
cyc = df[df['number_of_cyclist_injured'] + df['number_of_cyclist_killed'] > 0]
```

```python
df.to_csv("data/cleaned_collision_data.csv", index=False)
ped.to_csv("data/pedestrian_collisions.csv", index=False)
cyc.to_csv("data/cyclist_collisions.csv", index=False)
```
