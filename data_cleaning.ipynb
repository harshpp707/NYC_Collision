```python
import pandas as pd
import requests

```

```python
url = "https://data.cityofnewyork.us/resource/h9gi-nx95.json"
params = {"$limit": 50000} 
data = requests.get(url, params=params).json()
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
columns_needed = [
    "crash_date","crash_time","borough","latitude","longitude",
    "number_of_pedestrians_injured","number_of_pedestrians_killed",
    "number_of_cyclist_injured","number_of_cyclist_killed"
]
df = df[columns_needed]
df = df.dropna(subset=["borough", "latitude", "longitude"])
```

```python
injury_cols = [
    "number_of_pedestrians_injured","number_of_pedestrians_killed",
    "number_of_cyclist_injured","number_of_cyclist_killed"
]
```

```python
for col in injury_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0).astype(int)
```

```python
df = df[
    (df["number_of_pedestrians_injured"] > 0) |
    (df["number_of_pedestrians_killed"] > 0) |
    (df["number_of_cyclist_injured"] > 0) |
    (df["number_of_cyclist_killed"] > 0)
]
```

```python
df["crash_datetime"] = pd.to_datetime(
    df["crash_date"] + " " + df["crash_time"],
    format="%Y-%m-%d %H:%M",
    errors="coerce"
)
```
